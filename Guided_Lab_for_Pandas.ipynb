### Question 15 — Solution

```python
df_missing = df.copy()

df_missing.loc[2, "Rating"] = np.nan
df_missing.loc[6, "Rating"] = np.nan

print("Missing values:")
print(df_missing.isna())

print("\nMissing count by column:")
print(df_missing.isna().sum())

mean_rating = df_missing["Rating"].mean()
df_missing["Rating"] = df_missing["Rating"].fillna(mean_rating)

print("\nAfter filling missing ratings:")
print(df_missing[["Order_ID", "Rating"]])
```

### Question 16 — Solution

```python
# 1. Top 3 orders by sales
print("Top 3 orders:")
print(df.nlargest(3, "Sales")[["Order_ID", "Customer", "Sales"]])

# 2. Total sales by city
print("\nTotal sales by city:")
print(df.groupby("City")["Sales"].sum().sort_values(ascending=False))

# 3. Average rating by category
print("\nAverage rating by category:")
print(df.groupby("Category")["Rating"].mean())

# 4. Customer with highest total sales
customer_sales = df.groupby("Customer")["Sales"].sum().sort_values(ascending=False)
print("\nCustomer with highest total sales:")
print(customer_sales.head(1))

# 5. Orders after 28 February 2026
print("\nOrders after 28 February 2026:")
print(df[df["Order_Date"] > "2026-02-28"])

# 6. Create Sales_Level
conditions = [
    df["Sales"] >= 60000,
    df["Sales"].between(30000, 59999),
    df["Sales"] < 30000
]

labels = ["High", "Medium", "Low"]

df["Sales_Level"] = np.select(conditions, labels)
print("\nSales levels:")
print(df[["Order_ID", "Sales", "Sales_Level"]])
```

## Pandas Quick Reference

| Function / Method | Purpose |
|---|---|
| `pd.Series()` | Create a Series |
| `pd.DataFrame()` | Create a DataFrame |
| `head()` / `tail()` | View beginning / end of data |
| `shape` | Get rows and columns |
| `columns` | Get column names |
| `iloc[]` | Select by integer position |
| `df[condition]` | Filter rows |
| `sort_values()` | Sort data |
| `sum()` / `mean()` | Aggregate values |
| `groupby()` | Group data for analysis |
| `agg()` | Apply multiple aggregations |
| `value_counts()` | Count unique values |
| `isna()` | Detect missing values |
| `fillna()` | Fill missing values |
| `to_datetime()` | Convert values to dates |
| `nlargest()` | Get largest values |
