### Question 13 — Solution

```python
# 2x + 3y = 13
# 4x - y = 5

A = np.array([[2, 3],
              [4, -1]])

B = np.array([13, 5])

solution = np.linalg.solve(A, B)

x, y = solution

print("x =", x)
print("y =", y)
```

### Question 14 — Solution

```python
marks = np.array([72, 88, 91, 65, 78, 95, 84, 59])

average = np.mean(marks)
highest = np.max(marks)
lowest = np.min(marks)
count_75_plus = np.sum(marks >= 75)
above_average = marks[marks > average]
std_dev = np.std(marks)

print("Class average:", average)
print("Highest:", highest)
print("Lowest:", lowest)
print("Students scoring >= 75:", count_75_plus)
print("Marks above average:", above_average)
print("Standard deviation:", std_dev)
```

### Question 15 — Solution

```python
sales = np.array([120, 150, 135, 180, 165, 210, 195, 225])

print("Total sales:", np.sum(sales))
print("Average weekly sales:", np.mean(sales))
print("Highest sales:", np.max(sales))
print("Lowest sales:", np.min(sales))
print("Difference:", np.max(sales) - np.min(sales))
print("Shape:", sales.shape)
print("Size:", sales.size)
```

### Key NumPy functions used

| Function / Attribute | Purpose |
|---|---|
| `np.array()` | Create an array |
| `np.arange()` | Create evenly spaced values over a range |
| `np.zeros()` | Create an array filled with zeros |
| `.shape` | Get dimensions of an array |
| `.ndim` | Get number of dimensions |
| `.size` | Get total number of elements |
| `np.sum()` | Calculate sum |
| `np.mean()` | Calculate mean |
| `np.median()` | Calculate median |
| `np.min()` / `np.max()` | Find minimum / maximum |
| `np.std()` | Calculate standard deviation |
| `np.linalg.solve()` | Solve a system of linear equations |
