## 8. Generative AI with Python
### 8.1 Calling an LLM through an SDK
An LLM API takes **messages** in and returns generated text. Set your key as an environment variable (never hard-code it). The cell runs only if `ANTHROPIC_API_KEY` is set and the `anthropic` package is installed; the trainer will demo it live otherwise.

```python
import os
def ask_llm(prompt, system="You are a concise enterprise AI assistant."):
    try:
        import anthropic
        if not os.getenv("ANTHROPIC_API_KEY"):
            return "[No API key set: skipping live call. Set ANTHROPIC_API_KEY to try it.]"
        client = anthropic.Anthropic()
        r = client.messages.create(model="claude-sonnet-5", max_tokens=300, system=system,
                                   messages=[{"role": "user", "content": prompt}])
        return r.content[0].text
    except ImportError:
        return "[anthropic package not installed: %pip install anthropic]"

print(ask_llm("Explain churn prediction to a bank executive in 3 bullets."))
```

### 8.3 Mini-RAG: retrieval before generation
**Retrieval-Augmented Generation** finds the most relevant documents first, then passes them to the LLM as context, which reduces hallucination. Production systems use embeddings + vector databases; here we use TF-IDF so it runs anywhere.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

docs = [
    "Refunds are allowed within 30 days of purchase with a valid receipt.",
    "Premium support is available 24x7 for enterprise plan customers.",
    "Data is encrypted at rest and in transit using industry standards.",
    "Employees must complete AI ethics training every year.",
]
vec = TfidfVectorizer(stop_words="english").fit(docs)
D = vec.transform(docs)

def retrieve(question, k=1):
    sims = cosine_similarity(vec.transform([question]), D).ravel()
    return [docs[i] for i in sims.argsort()[::-1][:k]]

q = "How many days do I have to get a refund?"
ctx = retrieve(q)
rag_prompt = f"Answer ONLY using this context.\nContext: {ctx[0]}\nQuestion: {q}"
print(rag_prompt)
print("\nLLM answer:", ask_llm(rag_prompt))
```
