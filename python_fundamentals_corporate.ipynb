## 1. Variables and Data Types

**Theory:** A *variable* is a named container that stores a value. Python figures out the type automatically.

| Type | Meaning | Corporate example |
|---|---|---|
| `int` | Whole numbers | Number of employees: `250` |
| `float` | Decimal numbers | Quarterly revenue: `1250000.75` |
| `str` | Text | Employee name: `"Priya Sharma"` |
| `bool` | True / False | Is manager: `True` |

**Naming rules:** letters, digits, underscore; cannot start with a digit; no spaces; case-sensitive. Prefer `snake_case` such as `monthly_salary`.

````python
# Variables with different data types
employee_name = "Priya Sharma"     # str
employee_count = 250               # int
quarterly_revenue = 1250000.75     # float
is_manager = True                  # bool

print(employee_name, type(employee_name))
print(employee_count, type(employee_count))
print(quarterly_revenue, type(quarterly_revenue))
print(is_manager, type(is_manager))
````

### Arithmetic with `int` and `float`

````python
units_sold = 120          # int
unit_price = 49.99        # float
discount_rate = 0.10      # 10%

gross = units_sold * unit_price
discount = gross * discount_rate
net = gross - discount

print("Gross sales :", gross)
print("Discount    :", discount)
print("Net sales   :", round(net, 2))

# Other operators
print(17 // 5)   # floor division -> 3
print(17 % 5)    # remainder      -> 2
print(2 ** 10)   # power          -> 1024
````

### Type conversion (casting)
Data from files and forms often arrives as text. Convert it before doing maths.

````python
salary_text = "75000"
salary = int(salary_text)          # str -> int
bonus = float("2500.50")           # str -> float
print(salary + bonus)

print(str(2026) + " budget")       # int -> str
print(int(9.99))                   # float -> int (truncates!)
````

## 2. `print()` and `input()`

**Theory**
- `print()` displays output. Use **f-strings** (`f"...{variable}..."`) to build clear messages.

## 3. Strings

**Theory:** A string is a sequence of characters. Characters are **indexed** from `0`; negative indexes count from the end. Strings are **immutable** (cannot be changed in place; methods return new strings).

````python
email = "  Priya.Sharma@AcmeCorp.com  "

print(len(email))
clean = email.strip()              # remove outer spaces
print(clean)
print(clean.lower())               # lowercase
print(clean.upper())               # uppercase
````

### Indexing, slicing and useful methods

````python
emp_id = "EMP-2026-0457"

print(emp_id[0])        # 'E'
print(emp_id[-1])       # '7'
print(emp_id[0:3])      # 'EMP'   (start included, end excluded)
print(emp_id[4:8])      # '2026'
print(emp_id[::-1])     # reversed

print(emp_id.split("-"))               # ['EMP', '2026', '0457']
print(emp_id.replace("EMP", "CONTR"))  # replace text
print(emp_id.startswith("EMP"))        # True
print("2026" in emp_id)                # True (membership test)
````

````python
# Practical example: build a username and email from a full name
full_name = "priya sharma"
first, last = full_name.title().split()
username = (first[0] + last).lower()
print(f"Name: {first} {last} | Username: {username} | Email: {username}@acmecorp.com")

# Joining a list of strings
skills = ["Python", "SQL", "Excel"]
print(", ".join(skills))
````

## 4. Tuples

**Theory:** An ordered collection that is **immutable** (cannot be changed after creation). Written with `( )`. Use tuples for fixed data such as coordinates, fiscal quarters, or a record that must not change.

````python
fiscal_quarters = ("Q1", "Q2", "Q3", "Q4")
office_location = ("Pune", "India", 18.52)   # can mix types

print(fiscal_quarters[0])        # Q1
print(fiscal_quarters[-1])       # Q4
print(len(fiscal_quarters))
print("Q3" in fiscal_quarters)

# Unpacking
city, country, latitude = office_location
print(f"{city}, {country} (lat {latitude})")

# Immutability: this would raise an error
try:
    fiscal_quarters[0] = "Q5"
except TypeError as e:
    print("Error:", e)
````

## 5. Lists

**Theory:** An ordered, **mutable** collection written with `[ ]`. The workhorse of Python: store tasks, sales figures, employee names, and more.

| Method | Purpose |
|---|---|
| `append(x)` | add to the end |
| `insert(i, x)` | add at position |
| `remove(x)` | remove first match |
| `pop()` | remove & return last item |
| `sort()` | sort in place |

````python
monthly_sales = [42000, 38500, 51000, 47250]

monthly_sales.append(55000)          # add a new month
monthly_sales[1] = 39000             # correct an entry
print(monthly_sales)

print("Total  :", sum(monthly_sales))
print("Average:", sum(monthly_sales) / len(monthly_sales))
print("Best   :", max(monthly_sales))
print("Lowest :", min(monthly_sales))
````

````python
team = ["Asha", "Vikram", "Meera"]
team.append("Karan")
team.insert(1, "Neha")
team.remove("Vikram")
print(team)
print(sorted(team))        # returns a new sorted list
print(team[1:3])           # slicing works just like strings

# List comprehension: apply a 10% raise to all salaries
salaries = [50000, 62000, 71000]
new_salaries = [s * 1.10 for s in salaries]
print(new_salaries)
````

## 6. Dictionaries

**Theory:** Stores **key → value** pairs, written with `{ }`. Perfect for records such as an employee profile, a product, or a JSON response. Keys are unique; look up values by key, not by position.

````python
employee = {
    "id": "EMP-0457",
    "name": "Priya Sharma",
    "department": "Finance",
    "salary": 82000,
    "skills": ["Excel", "SQL"]
}

print(employee["name"])
print(employee.get("manager", "Not assigned"))   # safe lookup with default

employee["salary"] = 88000          # update
employee["location"] = "Pune"       # add new key
del employee["id"]                  # delete a key
print(employee)
````

````python
# Looping through a dictionary
dept_headcount = {"Finance": 25, "IT": 60, "HR": 12, "Sales": 80}

for dept, count in dept_headcount.items():
    print(f"{dept:<8} -> {count} employees")

print("Keys  :", list(dept_headcount.keys()))
print("Total :", sum(dept_headcount.values()))

# List of dictionaries = a mini table
staff = [
    {"name": "Asha", "salary": 60000},
    {"name": "Karan", "salary": 72000},
]
for person in staff:
    print(person["name"], person["salary"])
````

## 7. Sets

**Theory:** An **unordered** collection of **unique** items, written with `{ }` (an empty set is `set()`). Great for removing duplicates and comparing groups.

| Operation | Symbol | Meaning |
|---|---|---|
| Union | `a` &#124; `b` | in either set |
| Intersection | `a & b` | in both |
| Difference | `a - b` | in a but not b |

````python
# Remove duplicates
attendees = ["Asha", "Karan", "Asha", "Meera", "Karan"]
unique_attendees = set(attendees)
print(unique_attendees, len(unique_attendees))

# Compare two groups
python_team = {"Asha", "Karan", "Meera"}
sql_team = {"Karan", "Neha", "Meera"}

print("Both skills   :", python_team & sql_team)
print("Any skill     :", python_team | sql_team)
print("Python only   :", python_team - sql_team)

python_team.add("Rahul")
python_team.discard("Asha")
print(python_team)
````

## 8. Conditional Statements: `if`, `elif`, `else`

**Theory:** Code runs only when a condition is `True`. **Indentation (4 spaces) is mandatory** in Python.

Comparison operators: `==  !=  >  <  >=  <=`  |  Logical operators: `and`, `or`, `not`

````python
# Performance rating -> bonus
rating = 4.2
salary = 80000

if rating >= 4.5:
    bonus = salary * 0.20
    grade = "Outstanding"
elif rating >= 3.5:
    bonus = salary * 0.10
    grade = "Exceeds expectations"
elif rating >= 2.5:
    bonus = salary * 0.05
    grade = "Meets expectations"
else:
    bonus = 0
    grade = "Needs improvement"

print(f"Rating {rating}: {grade} | Bonus: {bonus:,.2f}")
````

````python
# Combining conditions: expense approval policy
amount = 1800
has_receipt = True
is_travel = True

if amount <= 2000 and has_receipt:
    print("Auto-approved")
elif amount > 2000 and has_receipt:
    print("Send to manager for approval")
else:
    print("Rejected: receipt required")

# Compact one-line form
status = "Eligible" if amount < 5000 else "Review"
print(status)
````

## 9. `for` Loop

**Theory:** Repeats code for each item in a sequence (list, string, dict, `range()`). Use it when you know *what* you are iterating over.

`range(5)` → 0,1,2,3,4 | `range(1, 6)` → 1..5 | `range(0, 10, 2)` → 0,2,4,6,8

````python
# Process invoices
invoices = [1200.50, 850.00, 2300.75, 410.25]
total = 0

for amount in invoices:
    total += amount

print(f"Total invoices: {total:,.2f}")
````

````python
# Loop with index and conditions
sales = {"North": 120000, "South": 85000, "East": 134000, "West": 67000}
target = 100000

for region, value in sales.items():
    if value >= target:
        print(f"✅ {region}: {value:,} (target met)")
    else:
        print(f"⚠️ {region}: {value:,} (short by {target - value:,})")

# range example: 5-year budget projection with 8% growth
budget = 500000
for year in range(1, 6):
    budget *= 1.08
    print(f"Year {year}: {budget:,.0f}")
````

````python
# break and continue
employees = ["Asha", "Karan", "", "Meera", "Neha"]
for name in employees:
    if name == "":
        continue          # skip blank records
    if name == "Meera":
        print("Found Meera, stopping search")
        break
    print("Checking", name)
````

## 10. `while` Loop

**Theory:** Repeats **as long as a condition stays `True`**. Use it when you don't know in advance how many repetitions are needed. Always make sure something inside the loop eventually makes the condition `False`, otherwise it runs forever.

````python
# How many months to reach a savings goal?
balance = 10000
goal = 25000
monthly_deposit = 1500
months = 0

while balance < goal:
    balance += monthly_deposit
    months += 1

print(f"Goal reached in {months} months with balance {balance:,}")
````

````python
# Input validation with while (typical real-world use)
# Run this cell and try entering a wrong value first.
attempts = 0
while attempts < 3:
    pin = input("Enter 4-digit approval code: ")
    if pin == "1234":
        print("Access granted")
        break
    attempts += 1
    print(f"Wrong code. Attempts left: {3 - attempts}")
else:
    print("Account locked. Contact IT support.")
````

## Practice Question 2: Expense Claim Processor

Employees submitted these expense claims as `(employee, amount, category)` tuples:

```python
claims = [
    ("Asha", 1200, "Travel"),
    ("Karan", 300, "Meals"),
    ("Asha", 4500, "Hotel"),
    ("Meera", 150, "Meals"),
    ("Karan", 9000, "Travel"),
]
```

**Tasks**
1. Use a `for` loop to build a **dictionary** of total claimed amount per employee.
2. Use a **set** to list the unique expense categories.
3. Flag any employee whose total exceeds the **10,000** limit (`if` statement).
4. Use a **`while` loop** to simulate a budget of **12,000**: deduct claims one by one from the list and stop when the remaining budget cannot cover the next claim. Print how many claims were paid.

<details>
<summary><b>👉 Click to show solution</b></summary>

```python
claims = [
    ("Asha", 1200, "Travel"),
    ("Karan", 300, "Meals"),
    ("Asha", 4500, "Hotel"),
    ("Meera", 150, "Meals"),
    ("Karan", 9000, "Travel"),
]

# 1. Total per employee (dictionary)
totals = {}
for employee, amount, category in claims:      # tuple unpacking
    totals[employee] = totals.get(employee, 0) + amount
print("Totals:", totals)

# 2. Unique categories (set)
categories = set()
for _, _, category in claims:
    categories.add(category)
print("Categories:", categories)

# 3. Flag employees over the limit
for employee, total in totals.items():
    if total > 10000:
        print(f"FLAG: {employee} exceeded limit with {total:,}")

# 4. Pay claims in order until the budget runs out
budget = 12000
i = 0
while i < len(claims) and budget >= claims[i][1]:
    budget -= claims[i][1]
    i += 1

print(f"Claims paid: {i} | Remaining budget: {budget:,}")
```
